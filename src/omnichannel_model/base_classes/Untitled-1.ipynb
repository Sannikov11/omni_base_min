```python
import pandas as pd



def calc_cr(df: pd.DataFrame) -> pd.DataFrame:
    grouped = df.groupby('group').agg(
        total_sessions=('sessions', 'sum'),
        total_conversions=('conversions', 'sum')
    )
    grouped['cr'] = grouped['total_conversions'] / grouped['total_sessions']
    return grouped.reset_index()[['group', 'cr']]

 
def flag_above_group_avg(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df['user_cr'] = df['conversions'] / df['sessions']
    group_avg = df.groupby('group')['user_cr'].transform('mean')
    df['above_avg'] = df['user_cr'] > group_avg
    return df


ab_df = pd.DataFrame({
"user_id": [1,2,3,4,5,6,7,8],
"group": ['A', 'A', 'B', 'B', 'A', 'B','A', 'B'],
"sessions": [3,5,2,4,1,6,2,3],
"conversions": [1,2,0,1,0,2,1,1]
})

```

```python

calc_cr(ab_df)
```

```python
flag_above_group_avg(ab_df)
```
